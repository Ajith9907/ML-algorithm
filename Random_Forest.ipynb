```python
import numpy as np
import pandas as pd
# Load dataset
df = pd.read_csv("titanic.csv")
df.head()
# Select required columns
df = df[["Pclass", "Sex", "Age", "Fare", "Survived"]]
```

```python
#Find missing values in Age column
print("Missing Age values before:", df["Age"].isnull().sum())

# Handle missing Age values
df["Age"] = df["Age"].fillna(df["Age"].median())

print("Missing Age values after:", df["Age"].isnull().sum())
```

```python
# Check the actual sex values
print(df["Sex"].unique())

df["Sex"] = df["Sex"].str.strip().str.lower()

#Convert sex to numbers
df["Sex"] = df["Sex"].map({
    "male": 0,
    "female": 1
})
print(df["Sex"].unique())
df.head()
```

```python
# Tree 1 - Sex
def tree_sex(sex):
    if sex == 1:       # Female
        return 1
    else:              # Male
        return 0

# Tree 2 - Pclass
def tree_pclass(pclass):
    if pclass <= 2:
        return 1
    else:
        return 0

# Tree 3 - Age
def tree_age(age):
    if age <= 15:
        return 1
    else:
        return 0

# Tree 4 - Fare
def tree_fare(fare):
    if fare >= 30:
        return 1
    else:
        return 0        

#create random forest to make predictions
def random_forest(pclass, sex, age, fare):

    prediction1 = tree_sex(sex)
    prediction2 = tree_pclass(pclass)
    prediction3 = tree_age(age)
    prediction4 = tree_fare(fare)

    votes = [
        prediction1,
        prediction2,
        prediction3,
        prediction4
    ]
    if sum(votes) >= 2:
        final_prediction = 1
    else:
        final_prediction = 0

    return final_prediction      
```

```python
# Test the first 10 passengers
predictions = []

for index in range(10):

    passenger = df.iloc[index]

    prediction = random_forest(
        passenger["Pclass"],
        passenger["Sex"],
        passenger["Age"],
        passenger["Fare"]
    )

    predictions.append(prediction)

    print(
        "Passenger", index + 1,
        "| Actual:", passenger["Survived"],
        "| Predicted:", prediction
    )
```

```python
# Calculate Accuracy
actual = df.iloc[:10]["Survived"].values

correct = sum(actual == np.array(predictions))

accuracy = (correct / 10) * 100

print("\nAccuracy:", accuracy, "%")
```

```python
# Test a new passenger
new_passenger = {
    "Pclass":1 ,
    "Sex": 0,       # male
    "Age": 30,
    "Fare": 40
}

prediction = random_forest(
    new_passenger["Pclass"],
    new_passenger["Sex"],
    new_passenger["Age"],
    new_passenger["Fare"]
)

print("\nNew Passenger Prediction")

if prediction == 1:
    print("Survived")
else:
    print("Did Not Survive")
```
