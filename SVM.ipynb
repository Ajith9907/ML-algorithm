```python
import pandas as pd
import numpy as np
# load csv file
df = pd.read_csv("mobile_phones_1000.csv")

print("First 5 rows:")
print(df.head())
```

```python
#Get input features
# x1 = RAM
# x2 = Price
X = df[["ram_gb", "price_thousand"]].values
```

```python
#get target
# 1 = Premium
# 0 = Budget
y = df["target"].values
#Convert target for svm
# Budget  = -1
# Premium = +1
y = np.where(y == 0, -1, 1)
#feature scaling
X_mean = X.mean(axis=0)
X_std = X.std(axis=0)
X_scaled = (X - X_mean) / X_std

```

```python
# Start w1, w2 and b 
w1 = 0.0
w2 = 0.0
b = 0.0

# Training Settings
learning_rate = 0.001  # Fixed typo here (was 'earning_rate')
C = 1
epochs = 1000

# Start training 
print("\n===============================")
print("STARTING SVM TRAINING")
print("===============================")

for epoch in range(epochs):
    for i in range(len(X_scaled)):
        x1 = X_scaled[i][0]
        x2 = X_scaled[i][1]
        
        decision = w1 * x1 + w2 * x2 + b
        margin = y[i] * decision
        
        if margin >= 1:
            w1 = w1 - learning_rate * (2 * w1)
            w2 = w2 - learning_rate * (2 * w2)
        else:
            w1 = w1 - learning_rate * (2 * w1 - C * y[i] * x1)
            w2 = w2 - learning_rate * (2 * w2 - C * y[i] * x2)
            b = b + learning_rate * C * y[i]
            
    # Properly indented inside the epoch loop (aligned with the inner for loop)
    if (epoch + 1) % 100 == 0:
        print("Epoch:", epoch + 1, "| w1 =", round(w1, 4), "| w2 =", round(w2, 4), "| b =", round(b, 4))
```

```python
#Training completed
print("\n======================================")
print("TRAINING COMPLETED")
print("======================================")

print("Learned w1 =", round(w1, 4))

print("Learned w2 =", round(w2, 4))

print("Learned b  =", round(b, 4))
```

```python
#Test the training data
predictions = []
for i in range(len(X_scaled)):
    x1 = X_scaled[i][0]

    x2 = X_scaled[i][1]
    decision = w1 * x1 + w2 * x2 + b
    if decision >= 0:
        prediction = 1   
    else:
        prediction = -1   
    predictions.append(prediction)
# Convert predictions into NumPy array
predictions = np.array(predictions)
# calculate accuracy
correct = np.sum(predictions == y)

total = len(y)

accuracy = correct / total


print("\n======================================")
print("MODEL ACCURACY")
print("======================================")
print("Correct predictions:", correct)
print("Total records:", total)
print("Accuracy:", round(accuracy * 100, 2), "%")

# TEST A COMPLETELY NEW PHONE

print("\n======================================")
print("TESTING A NEW PHONE")
print("======================================")
# New phone information
new_ram = 10
new_price = 50
print("RAM   :", new_ram, "GB")
print("Price : ₹", new_price, "thousand")

# SCALE THE NEW PHONE
# We must scale the new phone using
# the SAME mean and standard deviation
# used during training.

new_x1 = (new_ram - X_mean[0]) / X_std[0]

new_x2 = (new_price - X_mean[1]) / X_std[1]
print("\nScaled values:")
print("x1 =", round(new_x1, 4))
print("x2 =", round(new_x2, 4))

# APPLY THE SVM FORMULA
 
# We now use the w1, w2 and b
# learned during training.
# Formula:
# Decision = w1*x1 + w2*x2 + b

decision = (
    w1 * new_x1
    + w2 * new_x2
    + b
)
#SHOW THE ACTUAL CALCULATION
print("\n======================================")
print("SVM CALCULATION")
print("======================================")


print(
    "Decision = w1*x1 + w2*x2 + b"
)

print(
    "Decision =",
    round(w1, 4),
    "*",
    round(new_x1, 4),
    "+",
    round(w2, 4),
    "*",
    round(new_x2, 4),
    "+",
    round(b, 4)
)
print(
    "Decision =",
    round(decision, 4)
)

# FINAL PREDICTION
print("\n======================================")
print("FINAL PREDICTION")
print("======================================")
if decision >= 0:
    print("Prediction: PREMIUM PHONE")
else:
    print("Prediction: BUDGET PHONE")
```
