```python
#Problem: Predict house price based on size (Square Feet)
import numpy as np
import pandas as pd
#STEP 1: Create the dataset
data = {
    "Size": [1000, 1500, 2000, 2500, 3000],
    "Price": [150, 200, 250, 320, 380] # Price in $k
}
# Convert dictionary into a Pandas DataFrame
df = pd.DataFrame(data)
print("Dataset:")
print(df)
#
#STEP 2: Separate input (X) and output (y)
#
# X = Independent variable / Input
X = df["Size"].values
# y = Dependent variable / Output
y = df["Price"].values
print("\nX (Size):")
print(X)
print("\ny (Price):")
print(y)
#STEP 3: Calculate the mean
# Mean of X
x_mean = np.mean(X)
# Mean of y
y_mean = np.mean(y)
print("\nMean of X:", x_mean)
print("Mean of y:", y_mean)
#STEP 4: Calculate the slope (m)
# Formula:
# m = sum((x - x_mean) * (y - y_mean)) / sum((x - x_mean)^2)
# Slope tells us how much y changes when x increases by 1 unit.
numerator = np.sum((X - x_mean) * (y - y_mean))
denominator = np.sum((X - x_mean) ** 2)
m = numerator / denominator
print("\nSlope (m):", m)
#STEP 5: Calculate the intercept (b)
# Formula:
# b = y_mean - (m * x_mean)
b = y_mean - (m * x_mean)
print("Intercept (b):", b)
#STEP 6: Create the Linear Regression equation
print("\nLinear Regression Equation:")
print("Price =", m, "* Size +", b)
#STEP 7: Predict price for a new house
new_size = 2200
# Formula: y = mx + b
prediction = (m * new_size) + b
print("\nNew House Size:", new_size)
print("Predicted Price ($k):", prediction)
```
